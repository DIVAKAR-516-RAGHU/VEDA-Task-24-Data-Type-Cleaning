### Objective
Identify incorrect data types in a dataset and convert columns to appropriate data types using Pandas.

### Description
Data type cleaning is the process of identifying and correcting inappropriate data types in a dataset to ensure accurate data analysis and machine learning.

### Tools Used

- Pandas
- Jupyter Notebook

### Dataset
Employee Dataset

### Deliverables
- Original data types
- Corrected data types
- Explanation of data type conversions

### Learning Outcome
Learned how to inspect data types using `dtypes`, convert columns using `pd.to_numeric()` and `pd.to_datetime()`, and verify the cleaned dataset.

```python
import pandas as pd
```

```python
data = {
    "Employee_ID": ["101", "102", "103", "104", "105"],
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": ["25", "30", "28", "35", "27"],
    "Salary": ["25000", "35000", "30000", "45000", "28000"],
    "Joining_Date": [
        "2024-01-15",
        "2023-06-20",
        "2024-03-10",
        "2022-11-05",
        "2023-09-18"
    ]
}

df = pd.DataFrame(data)

df
```

```python
print("Original Data Types:")
print(df.dtypes)
```

```python
df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print("Data types after conversion:")
print(df.dtypes)
```

```python
print("Cleaned Dataset:")
display(df)

print("\nFinal Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
comparison = pd.DataFrame({
    "Original Data Type": pd.DataFrame(data).dtypes,
    "Corrected Data Type": df.dtypes
})

comparison
```

```python
print("Average Age:", df["Age"].mean())
print("Average Salary:", df["Salary"].mean())

print("\nEmployees who earn more than 30000:")
display(df[df["Salary"] > 30000])
```
